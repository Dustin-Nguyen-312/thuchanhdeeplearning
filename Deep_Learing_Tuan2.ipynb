```python
import pandas as pd

#Đọc file dataset
df = pd.read_csv('housingdataset.csv')

print("Các cột có trong dataset này là: ")
print(df.columns.tolist())

print("\n10 dòng đầu tiên của dữ liệu: ")
print(df.head(10))

#Làm sạch dataset
print("Số lượng dữ liệu bị thiếu trong từng cột: ")
print(df.isnull().sum())

#Xóa dòng thiếu thông tin
df_clean = df.dropna()

print(f"\n Số lượng dòng ban đầu: {len(df)}")
print(f"Số lượng dòng sau khi làm sạch: {len(df_clean)}")

#Cập nhật lại df
df = df_clean

#Chọn lọc đặc trưng (Feature Selection)
features = ['total_rooms', 'ocean_proximity', 'total_bedrooms', 'housing_median_age', 'population', 'households', 'median_income']
target = 'median_house_value'

#Tạo dataframe mới
df_model = df[features + [target]].copy()

#Mã hóa dữ liệu (One-hot Encoding)
df_model = pd.get_dummies(df_model, columns=['ocean_proximity'], drop_first=True, dtype=int)

#Tách dữ liệu đầu vào
X = df_model.drop(columns=[target])
Y = df_model[target]

#Chia tập dữ liệu 80% học, 20% dự đoán
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

print(f"Số lượng căn nhà mang đi huấn luyện (Train): {X_train.shape[0]}")
print(f"Số lượng căn nhà mang đi dự đoán (Test): {X_test.shape[0]}")
print("\n Bảng dữ liệu đầu vào X (sau khi mã hóa thành số) là: ")
print(X_train.head(3))

```
